```python
class Bank:
  def __init__(self,name,money):
    self.name=name
    self.money=money
  def withdraw(self):
    num=int(input("請輸入提款金額:"))
    if num<=self.money:
      self.money=self.money-num
      print("帳戶名稱:",self.name)
      print("剩餘餘額:",self.money)
    else:
      print("餘額不足")

user=Bank("小明",10000)
user.withdraw()
```
