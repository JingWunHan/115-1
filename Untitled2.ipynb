```python
def calculate_bmi(height, weight):
    """
    計算並回傳 BMI 值。
    height: 身高 (公分)
    weight: 體重 (公斤)
    """
    bmi = weight / (height / 100) ** 2
    return bmi

# 第一組數據
h1, w1 = 170, 60
bmi1 = calculate_bmi(h1, w1)
print(f"BMI 1: {bmi1}")

# 第二組數據
h2, w2 = 165, 55
bmi2 = calculate_bmi(h2, w2)
print(f"BMI 2: {bmi2}")

# 第三組數據
h3, w3 = 180, 80
bmi3 = calculate_bmi(h3, w3)
print(f"BMI 3: {bmi3}")
```
