```python
order = []
print("請輸入您要點的餐點, 輸入'/'結束: ")
while True:
  item = input()
  if item == "/":
    break
  order.append(item)
  if item == "沙拉":
    order.remove("沙拉")
    print("今日無沙拉,已自動刪除")
  else:
    order.append("冰淇淋")

  print(f"完整訂單清單: {','.join(order)}")
```
